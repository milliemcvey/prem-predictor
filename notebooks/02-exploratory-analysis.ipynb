VISUALISATION SETUP:

```python
#FETCHES COLUMNS FROM DATASET

#Imports
import pandas as pd
import matplotlib.pyplot as plt

#Get column names
df = pd.read_csv("../data/cleaned/epl-dataset-clean.csv")
print(df.columns)

#Fetches Avg. / Min/Max / Standard Deviation
df.describe()
```

EXPLORATORY VISUALISATIONS:

```python
#VIS01: Points vs. Final League Standing

plt.figure(figsize=(10,5))

#Scatter Plot 'Pos' vs. 'Pts'
plt.scatter(df["Pos"], df["Pts"])
plt.title("Points vs. Final League Standing")
plt.xlabel("Position")
plt.ylabel("Points")

plt.show()
```

```python
#VIS02: 2000-2026 Total Wins per Team

#Fetch each winning team & Count
winning_teams = df[df["Pos"]== 1]
titles = winning_teams["Team"].value_counts()

print(titles)

#Create Graph
plt.figure(figsize=(10,5))

titles.plot(kind="bar")
plt.title("Total Wins per Team")
plt.xlabel("Team")
plt.ylabel("Number of League Titles")

plt.xticks(fontsize=8)
plt.xticks(rotation=0)

plt.show()
```

```python
#VIS03: 2000-2026 Points to Win EPL By Year

#Fetch each winning team's total points each year
winners = df[df["Pos"]== 1]
#Group by year
points_to_win = winners.groupby("Season")["Pts"].sum()

#Plot graph
plt.figure(figsize=(10,5))

plt.plot(points_to_win.index, points_to_win.values)

plt.title("Points Needed to Win the Premier League by Year")
plt.xlabel("Season")
plt.ylabel("Winning Points")

plt.xticks(rotation=90, fontsize=8)

plt.show()
```

```python
#VIS04: 2000-2026 Average Goal Difference per Team 

#Calculate Average Goal Difference
avg_gd = df.groupby("Team")["GD"].mean()

print(avg_gd)

#Plot bar chart
plt.figure(figsize=(10,5))

avg_gd.sort_values(ascending=False).plot(kind="bar")

plt.title("Average Goal Difference by Team")
plt.xlabel("Team")
plt.ylabel("Average Goal Difference")

plt.xticks(rotation=90, fontsize=8)
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

plt.show()
```

```python
#VIS05: 2000-2026 Average Points per Team

#Calculate Average Points
avg_pts = df.groupby("Team")["Pts"].mean()

print(avg_pts)

#Plot bar chart
plt.figure(figsize=(10,5))

avg_pts.sort_values(ascending=False).plot(kind="bar")

plt.title("Average Points by Team")
plt.xlabel("Team")
plt.ylabel("Average Points")

plt.xticks(rotation=90, fontsize=8)
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

plt.show()
```

```python
#VIS06: 2000-2026 Average League Position per Team

#Calculate Average Position
avg_pos = df.groupby("Team")["Pos"].mean()

print(avg_pos)

#Plot bar chart
plt.figure(figsize=(10,5))

avg_pos.sort_values(ascending=False).plot(kind="bar")

plt.title("Average Position by Team")
plt.xlabel("Team")
plt.ylabel("Average Position")

plt.xticks(rotation=90, fontsize=8)
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

plt.show()
```

```python
#VIS07: 2000-2026 Average Goals per Team

#Calculate Average Goals
avg_gf = df.groupby("Team")["GF"].mean()

print(avg_gf)

#Plot bar chart
plt.figure(figsize=(10,5))

avg_gf.sort_values(ascending=False).plot(kind="bar")

plt.title("Average Goals by Team")
plt.xlabel("Team")
plt.ylabel("Average Goals")

plt.xticks(rotation=90, fontsize=8)
plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

plt.show()
```

```python
#VIS08: 2000-2026 Win Rate (%) per Team

#Total win/losses by team (all seasons)
team_wl = df.groupby("Team")[["W", "L"]].sum()

#Percentage Calculations

team_wl["WinPct"] = team_wl["W"] / 38
team_wl["LossPct"] = team_wl["L"] / 38

#Plot 
team_wl = team_wl.sort_values(by="WinPct", ascending=False)
plt.figure(figsize=(12,6))

plt.bar(team_wl.index, team_wl["WinPct"], label="Win %")
plt.bar(team_wl.index, team_wl["LossPct"],
        bottom=team_wl["WinPct"],
        label="Loss %")

plt.title("Win vs Loss Percentage by Team")
plt.xlabel("Team")
plt.ylabel("Proportion")

plt.xticks(rotation=90)
plt.legend()

plt.show()
```

```python
#VIS09: 2000-2026 EPL Final Position Concistency per Team (Histogram)

team_order = df.groupby("Team")["Pos"].median().sort_values().index

df["Team"] = pd.Categorical(df["Team"], categories=team_order, ordered=True)

fig, ax = plt.subplots(figsize=(12,16))

df.boxplot(
    column="Pos",
    by="Team",
    vert=False,
    widths=0.6,
    ax=ax
)

plt.title("EPL Final Position Consistency by Team (2000–2026)")
plt.suptitle("")

plt.xlabel("Final League Position")
plt.ylabel("Team")

plt.xticks(fontsize=8)
plt.yticks(fontsize=8)

plt.gca().invert_xaxis()

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.show()
```

```python
#VIS10: Most EPL Appearances Since 2000

# Count appearances
teams_count = df["Team"].value_counts()

print(teams_count)

# Create Graph
plt.figure(figsize=(10,5))

teams_count.plot(kind="bar")

plt.title("Total EPL Appearances per Team")
plt.xlabel("Team")
plt.ylabel("Number of League Appearances")

plt.xticks(fontsize=8, rotation=90)

plt.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

plt.show()
```
